```python
import pandas as pd
```

```python
df = pd.read_csv('MH_Aadhar_Enrolment_Data.csv')
df.head()
```

```python
df.columns
```

```python
df.shape
```

```python
mh_df = df[df['state']=='Maharashtra'].copy()
mh_df.shape
```

```python
mh=df['district'].nunique()
mh_df['district'].value_counts().head(10)
```

```python
mh_df["date"] = pd.to_datetime(mh_df["date"], dayfirst=True)
mh_df["month"] = mh_df["date"].dt.to_period("M").astype(str)
```

```python
mh_df[["date", "month"]].head()
```

```python
city_month = (
    mh_df
    .groupby(["district", "month"], as_index=False)
    [["age_0_5", "age_5_17", "age_18_greater"]]
    .sum()
)

city_month.head()
city_month.shape
```

```python
age_long = city_month.melt(
    id_vars=["district", "month"],
    value_vars=["age_0_5", "age_5_17", "age_18_greater"],
    var_name="age_group",
    value_name="count"
)
```

```python
age_long.head()
```

```python
age_long.shape
```

```python
age_long["age_group"].unique()
```

```python
age_long["district"].nunique()
```

```python
age_long["month"].nunique()
```

```python
age_month = (
    age_long
    .groupby(["month", "age_group"], as_index=False)["count"]
    .sum()
)

age_month.head()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

for age in age_month["age_group"].unique():
    subset = age_month[age_month["age_group"] == age]
    plt.plot(subset["month"], subset["count"], marker="o", label=age)

plt.xlabel("Month")
plt.ylabel("Total Enrolments")
plt.title("Monthly Aadhaar Enrolment Trends by Age Group (Maharashtra)")
plt.legend(title="Age Group")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
city_month["total_activity"] = (
    city_month["age_0_5"]
    + city_month["age_5_17"]
    + city_month["age_18_greater"]
)
```

```python
city_month[["district", "month", "total_activity"]].head()
```

```python
enrolment_city_month = (
    city_month
    .groupby(["district", "month"], as_index=False)["total_activity"]
    .sum()
)
```

```python
heatmap_data = enrolment_city_month.pivot(
    index="district",
    columns="month",
    values="total_activity"
).fillna(0)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 8))
sns.heatmap(heatmap_data, cmap="YlOrRd")
plt.title("City-wise Aadhaar Enrolment Intensity Over Time (Maharashtra)")
plt.xlabel("Month")
plt.ylabel("District")
plt.tight_layout()
plt.show()
```

```python
demo_df = pd.read_csv('MH_Aadhar_Demographic_Data.csv')
demo_df = demo_df[demo_df['state']== 'Maharashtra'].copy()
```

```python
demo_df["date"] = pd.to_datetime(demo_df["date"], dayfirst=True)
demo_df["month"] = demo_df["date"].dt.to_period("M").astype(str)
```

```python
demo_city_month = (
    demo_df
    .groupby(["district", "month"], as_index=False)
    [["demo_age_5_17", "demo_age_17_"]]
    .sum()
)
```

```python
demo_city_month["total_activity"] = (
    demo_city_month["demo_age_5_17"]
    + demo_city_month["demo_age_17_"]
)
```

```python
demo_age_long = demo_city_month.melt(
    id_vars=["district", "month"],
    value_vars=["demo_age_5_17", "demo_age_17_"],
    var_name="age_group",
    value_name="count"
)
```

```python
demo_age_month = (
    demo_age_long
    .groupby(["month", "age_group"], as_index=False)["count"]
    .sum()
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
for age in demo_age_month["age_group"].unique():
    subset = demo_age_month[demo_age_month["age_group"] == age]
    plt.plot(subset["month"], subset["count"], marker="o", label=age)

plt.title("Monthly Aadhaar Demographic Update Trends by Age Group (Maharashtra)")
plt.xlabel("Month")
plt.ylabel("Total Updates")
plt.legend(title="Age Group")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns

demo_heatmap_data = (
    demo_city_month
    .pivot(index="district", columns="month", values="total_activity")
    .fillna(0)
)

plt.figure(figsize=(12, 8))
sns.heatmap(demo_heatmap_data, cmap="Blues")
plt.title("City-wise Aadhaar Demographic Update Intensity Over Time (Maharashtra)")
plt.xlabel("Month")
plt.ylabel("District")
plt.tight_layout()
plt.show()
```
